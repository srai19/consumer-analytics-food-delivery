```python
import pandas as pd
# Load dataset
df = pd.read_csv("/content/ecommerce_customer_behavior_dataset_v2.csv")

# Revenue overview metrics
result = pd.DataFrame({
    "Metric": ["Total Orders", "Total Customers", "Total Revenue", "Average Order Value"],
    "Value": [
        df["Order_ID"].nunique(),
        df["Customer_ID"].nunique(),
        df["Total_Amount"].sum(),
        df["Total_Amount"].mean()
    ]
})

result
```

### Business Insight

- Establishes the baseline revenue and order volume of the platform.
- Average order value provides pricing context for profitability analysis.
- This baseline will be used to compare retention and discount impact in subsequent analyses.
