```python
import pandas as pd

# Load data
df = pd.read_csv("/content/ecommerce_customer_behavior_dataset_v2.csv")

# Retention vs profitability analysis
retention_summary = (
    df.groupby("Is_Returning_Customer")
      .agg(
          customers=("Customer_ID", "nunique"),
          total_orders=("Order_ID", "nunique"),
          total_revenue=("Total_Amount", "sum"),
          avg_order_value=("Total_Amount", "mean"),
          total_discount_given=("Discount_Amount", "sum")
      )
      .reset_index()
)

retention_summary
```

### Business Insights

- Returning customers generate higher revenue per order compared to first-time customers.
- Order frequency is higher for returning customers, indicating stronger engagement.
- A significant portion of discounts is consumed by returning customers, suggesting potential discount dependency.
- Retention strategies should focus on experience and loyalty rather than blanket discounting.
